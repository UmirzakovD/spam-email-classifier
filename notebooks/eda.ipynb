# Exploratory Data Analysis - SMS Spam Collection

Quick look at the dataset used by the Spam Email Classifier project: class balance, message length distribution, and the most frequent words per class.

Run the first cell before anything else - it downloads/loads the dataset via `src.data_loader`.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import matplotlib.pyplot as plt
import pandas as pd

from src.data_loader import load_dataset
from src.preprocessing import clean_corpus

df = load_dataset()
df.head()
```

```python
print(f"Total messages: {len(df)}")
print(df['label'].value_counts())
print(df['label'].value_counts(normalize=True).round(3))
```

## Class balance

```python
counts = df['label'].value_counts()
plt.figure(figsize=(5, 4))
plt.bar(counts.index, counts.values, color=['#4C72B0', '#DD8452'])
plt.title('Class distribution (ham vs spam)')
plt.ylabel('Number of messages')
plt.show()
```

## Message length by class

```python
df['length'] = df['text'].str.len()
df.groupby('label')['length'].describe()
```

```python
plt.figure(figsize=(7, 4))
for label, color in [('ham', '#4C72B0'), ('spam', '#DD8452')]:
    subset = df.loc[df['label'] == label, 'length']
    plt.hist(subset, bins=40, alpha=0.6, label=label, color=color)
plt.xlabel('Message length (characters)')
plt.ylabel('Frequency')
plt.title('Message length distribution by class')
plt.legend()
plt.show()
```

## Most frequent words after cleaning

```python
df['clean_text'] = clean_corpus(df['text'])

from collections import Counter

def top_words(label, n=15):
    tokens = ' '.join(df.loc[df['label'] == label, 'clean_text']).split()
    return Counter(tokens).most_common(n)

print('Top spam words:', top_words('spam'))
print()
print('Top ham words:', top_words('ham'))
```

```python
spam_words = pd.DataFrame(top_words('spam'), columns=['word', 'count'])
plt.figure(figsize=(7, 5))
plt.barh(spam_words['word'][::-1], spam_words['count'][::-1], color='#DD8452')
plt.title('Top 15 words in spam messages (cleaned)')
plt.xlabel('Count')
plt.show()
```
